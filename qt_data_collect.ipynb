```python
import pandas as pd
import yfinance as yf
from pandas_datareader import data as pdr
```

```python
#get daily OHLCV

def get_ohlcv(ticker, start_date, end_date):
    df = yf.download(
        ticker,
        start=start_date,
        end=end_date,
        auto_adjust=False,
        progress=False
    )

    return df
```

```python
#get daily interest rate

def get_fred_data(start_date, end_date):

    series = {
    "DGS2": "US2Y",
    "DGS5": "US5Y",
    "DGS10": "US10Y",
    "DGS30": "US30Y",
    "FEDFUNDS": "FEDFUNDS",
    "DEXKOUS": "USDKRW"
    }

    data = pdr.DataReader(
        list(series.keys()),
        "fred",
        start_date,
        end_date
    )

    data = data.rename(columns=series)

    return data


```

```python
qqq = get_ohlcv(
    "QQQ",
    "2024-01-01",
    "2026-01-01"
)

print(qqq)
```

```python
df = get_fred_data(
    "2024-01-01",
    "2026-01-01"
)

print(df.head(20))
```
